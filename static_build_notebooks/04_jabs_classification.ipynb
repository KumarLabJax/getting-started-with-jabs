## The Active-Learning Loop

**Label, train, predict, refine.** You do not label an entire video up front. You label a
modest number of example frames for your behavior, train a classifier on them, then use the
classifier's own predictions to find where it disagrees with you - that disagreement is exactly
where labeling next teaches the classifier the most. Each pass through the loop needs less new
labeling than the last.

Concretely, inside a JABS project:

1. **Label** - watch a video in the JABS window, mark frames as your behavior or not
2. **Train** - JABS fits a classifier (this course uses **XGBoost**; JABS also supports Random
   Forest, its default, and CatBoost - see [JABS documentation, User Guide: Classifier
   Types](https://kumarlabjax.github.io/JABS-behavior-classifier/user-guide/classifier-types/))
   on kinematic features computed from the pose keypoints over your labeled frames
3. **Predict** - run the trained classifier on the rest of the video (or other videos in the
   project)
4. **Refine** - compare the classifier's predictions against your own labels, find the frames
   where they disagree, label there next, and retrain

That loop is why JABS calls itself an active-learning system: the classifier tells you where it
is uncertain, and that is where your next labeling effort goes furthest.

## Opening a Project

> "A JABS project is a directory of video files and their corresponding pose estimation files."
> - [JABS documentation, User Guide: Project Setup](https://kumarlabjax.github.io/JABS-behavior-classifier/user-guide/project-setup/)

Concretely, that's the same `data/` layout used throughout this course's QC and ingestion pages
- videos, pose files, and any existing annotations in one folder. The first time JABS opens a
project directory, it creates a `jabs/` subdirectory to hold project state: settings and
metadata (`project.json`), your labels, generated features, trained classifiers, and the latest
predictions.

Launch JABS pointed at that folder from the command line:

```bash
jabs /path/to/your/project
```

JABS opens with the project's videos listed in a sidebar. Select one to start reviewing frames
and labeling. The [JABS Demo](live_demo.html) page picks up from exactly this
point, using the QC-passed sessions from the previous page as the project.
