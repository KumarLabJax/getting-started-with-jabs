## What Is JABS?

**JABS** (the JAX Animal Behavior System) is an open-source tool from the [Kumar Lab](https://www.kumarlab.org/) for **labeling and classifying animal behavior from pose data**. Instead of scoring hours of video by hand, you annotate a modest number of example frames for a behavior of interest, and JABS trains a machine-learning classifier to find that behavior across the rest of your data.

> "JABS is a machine learning-based tool for automatically identifying and classifying animal behaviors in video recordings... it uses pose estimation data to extract features and train classifiers that can recognize specific behaviors across multiple animals and videos."
> - [JABS documentation, User Guide: Overview](https://kumarlabjax.github.io/JABS-behavior-classifier/user-guide/overview/)

**JABS is an active learning system.** You label a few examples, train, see where the classifier disagrees with you, label there next, and retrain - each round needs less labeling than the last. That loop, not a one-shot "label everything up front," is what makes it practical to build a working classifier from a laptop.

The pipeline this module follows:

```
video  ->  pose estimation  ->  feature extraction  ->  behavior classifier  ->  behavior labels
              (keypoints)          (JABS)                (this course: XGBoost)     (ethograms)
```

JABS supports three classifier algorithms - it does not require a deep neural network. Per the [JABS documentation, User Guide: Classifier Types](https://kumarlabjax.github.io/JABS-behavior-classifier/user-guide/classifier-types/):

> "Random Forest is the default classifier and a good starting point for most users."
> "CatBoost is a gradient boosting algorithm that can achieve excellent performance, particularly for complex behaviors."
> "XGBoost is another gradient boosting algorithm known for winning machine learning competitions."

This course's classifiers use **XGBoost**, chosen for this dataset and behavior - not because it's JABS's default. All three are tree-based models: they work well on the structured features JABS computes from keypoints (distances, angles, speeds), train quickly on modest hardware, and need far less labeled data than deep models - all of which matter for a scientist building a classifier on a laptop.

This module walks that pipeline using one curated example dataset. If you have your own open-field-arena recordings and want to run them through the same pipeline, see the [Your Own Data!](deep_dive_your_data.html) deep dive.

You will use the JABS GUI yourself to annotate a behavior and train a classifier in the live hands-on workflow - either live, in person on a course VM, or async and self-paced (see Module Overview below).

> **A first note on defining behavior.** Before you can teach a computer to recognize a behavior, you have to define what that behavior *is*. That sounds simple - until you try to write down exactly when `grooming` starts and ends, whether a mouse sniffing its own paw counts, and whether two people watching the same video would agree. How clearly you define a behavior before labeling largely determines how well your classifier works. We will return to this when you annotate and train your own classifier.

## Environment Setup

This page needs no setup - you are reading precomputed output. Setup only matters for the live
hands-on workflow, and what you need to do depends on which path you are on.

**Live, in person:** your course VM comes with the environment already set up and JABS
preinstalled. There is nothing to do here.

**Async, self-paced:** you will set up the environment yourself. This module pins specific
package versions (see `requirements.txt`), and its content is tied to that pinned JABS
environment - if you are using JABS more generally beyond this module, the
[JABS tutorial](https://jabs-tutorial.readthedocs.io/en/latest/) and the
[JABS-behavior-classifier repo](https://github.com/KumarLabJax/JABS-behavior-classifier) are the
canonical, always-current references.

```bash
# Create a virtual environment
python -m venv venv

# Activate it
# Windows:
venv\Scripts\activate
# macOS / Linux:
source venv/bin/activate

# Install all required packages
pip install -r requirements.txt
```

Then start JupyterLab from the same activated terminal:

```bash
jupyter lab
```

### Common failures (async setup)

| Problem | Fix |
|---------|-----|
| A package shows `[MISSING]` | `pip install -r requirements.txt` in the active environment |
| A package shows `[OUTDATED]` | `pip install --upgrade <package-name>` |
| All packages `[MISSING]` | Virtual environment is not activated - run the activate command above |
| `jabs` shows `[MISSING]` after install | Try `pip install jabs-behavior-classifier==0.47.1` directly |
